```python
centers = [[2, 2], [-2, -2], [2, -2]]
X, labels_true = make_blobs(n_samples=3000, centers=centers, cluster_std=0.9)

from sklearn.cluster import KMeans
n_clusters=3
k_means = KMeans(init="k-means++", n_clusters=n_clusters, random_state = 808 , n_init = 'auto'  )
k_means.fit(X)
print(k_means.cluster_centers_)
print("score", k_means.score(X))

from sklearn.metrics import silhouette_score
k_means_labels = k_means.predict(X)
print("silhouette_score: ", silhouette_score(X,k_means_labels ))
print("silhouette_score: ", silhouette_score(X,labels_true ))


fig = plt.figure(figsize=(12, 6))
colors = ["#4EACC5", "#FF9C34", "#4E9A06","#111111"]
colors = ["#FF0000", "#0000FF", "#00FF00","#111111"]

# KMeans
ax = fig.add_subplot(1, 2, 1)

# for k, col in zip(range(n_clusters), colors):
for k in range(n_clusters):
        ax.plot(X[labels_true == k, 0], X[labels_true == k, 1], "w", markerfacecolor=colors[k], marker="o", markersize=6, alpha = 1)

ax.spines['bottom'].set_position('zero')
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_linestyle('dotted')
ax.spines['left'].set_linestyle('dotted')

# Customize the appearance of grid lines (dotted and alpha=0.5)
ax.xaxis.grid(True, linestyle='--', alpha=0.5)
ax.yaxis.grid(True, linestyle='--', alpha=0.5)
# Remove the top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_title("Original")

ax = fig.add_subplot(1, 2, 2)
for k in range(n_clusters):
        cluster_center = k_means.cluster_centers_[k]
        ax.plot(X[k_means_labels == k, 0], X[k_means_labels == k, 1], "w", markerfacecolor=colors[k], marker="o", markersize=6, alpha = 1)
for k in range(n_clusters):
        cluster_center = k_means.cluster_centers_[k]
        ax.plot(
            cluster_center[0],
            cluster_center[1],
            "o",
            markerfacecolor='#000',
            markeredgecolor=colors[k],
            markersize=9,
        )
ax.spines['bottom'].set_position('zero')
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_linestyle('dotted')
ax.spines['left'].set_linestyle('dotted')

# Customize the appearance of grid lines (dotted and alpha=0.5)
ax.xaxis.grid(True, linestyle='--', alpha=0.5)
ax.yaxis.grid(True, linestyle='--', alpha=0.5)
# Remove the top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_title("KMeans")
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import silhouette_score
k_means_labels = k_means.predict(X)
print("silhouette_score: ", silhouette_score(X,k_means_labels ))
```

```python
centers = [[2, 2], [-2, -2], [2, -2]]
X, labels_true = make_blobs(n_samples=3000, centers=centers, cluster_std=0.9)

silhouette_scores = []
for n_clusters in range(2,8):
        print()
        print("n_clusters",n_clusters)
        k_means = KMeans(init="k-means++", n_clusters=n_clusters, random_state = 808 , n_init = 'auto'  )
        k_means.fit(X)
        print("score", k_means.score(X))

        k_means_labels = k_means.predict(X)
        print("silhouette_score: ", silhouette_score(X,k_means_labels ))
        silhouette_scores.append(silhouette_score(X,k_means_labels ))

fig = plt.figure(figsize=(9, 6))

ax = fig.add_subplot(1, 1, 1)
plt.plot(range(2,8), silhouette_scores)
ax.set_title('silhouette scores')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_xlabel('Nombre de clusters')
ax.set_ylabel('Coefficient de silhouette')
plt.grid()
plt.show()
```

```python
from sklearn import datasets
iris = datasets.load_iris()
X = iris.data
y = iris.target
```

```python
model = KMeans(n_clusters=3, n_init="auto")
model.fit(X)
```

```python
# Voici les scores de classification
print("score", model.score(X))
print("silhouette_score: ", silhouette_score(X,model.labels_ ))
```

```python

from sklearn.metrics import confusion_matrix

labels = model.labels_.copy()
print("confusion_matrix\n",confusion_matrix(y,labels ))
print(set(labels))
```

```python
# Match categories de kmeans avec les bonnes  categories du dataset
labels = model.labels_.copy()
# temp values
labels[labels==0] = 5
labels[labels==1] = 10
labels[labels==2] = 15

# Bonnes catégories
labels[labels == 5] = 0
labels[labels == 10] = 2
labels[labels == 15] = 1
```

```python
print("confusion_matrix\n",confusion_matrix(y,labels ))
```

```python
from sklearn.metrics import accuracy_score

print("accuracy_score",accuracy_score(y,labels ))
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# === Sélection de 3 features pour correspondre au graphe ===
# (comme sur ton image : largeur pétales, longueur sépales, longueur pétales)
X_plot = X[:, [3, 0, 2]]  # petal width, sepal length, petal length

# === Création de la figure ===
fig = plt.figure(figsize=(12, 5))

# =========================
# 1. CLUSTERS KMEANS
# =========================
ax1 = fig.add_subplot(121, projection='3d')

ax1.scatter(
    X_plot[:, 0],  # largeur pétales
    X_plot[:, 1],  # longueur sépales
    X_plot[:, 2],  # longueur pétales
    c=model.labels_,
    cmap='viridis',
    s=40
)

ax1.set_title("3 clusters")
ax1.set_xlabel("Largeur des pétales")
ax1.set_ylabel("Longueur des sépales")
ax1.set_zlabel("Longueur des pétales")


# =========================
# 2. DONNÉES RÉELLES
# =========================
ax2 = fig.add_subplot(122, projection='3d')

scatter = ax2.scatter(
    X_plot[:, 0],
    X_plot[:, 1],
    X_plot[:, 2],
    c=y,
    cmap='viridis',
    s=40
)

ax2.set_title("Données réelles")
ax2.set_xlabel("Largeur des pétales")
ax2.set_ylabel("Longueur des sépales")
ax2.set_zlabel("Longueur des pétales")

# Ajout des labels texte (approximation des centres)
centers = [
    X_plot[y == i].mean(axis=0)
    for i in range(3)
]

names = ["Setosa", "Versicolour", "Virginica"]

for center, name in zip(centers, names):
    ax2.text(center[0], center[1], center[2], name, fontsize=10)

plt.tight_layout()
plt.show()
```
