```python
col = 'Emission_GES_kgCO2/m²/an'
df = df[~df[col].isna()].copy()
df.reset_index(inplace = True, drop = True)
```

```python
from scipy import stats

col = 'Emission_GES_kgCO2/m²/an'

# seuil 3
df_z3 = df[df['z_emissions'].abs() <= 3]
nb_supprimes_3 = len(df) - len(df_z3)

# seuil 2
df_z2 = df[df['z_emissions'].abs() <= 2]
nb_supprimes_2 = len(df) - len(df_z2)
print(f"  Supprimés : {nb_supprimes_3}")
print(f"  Supprimés : {nb_supprimes_2}")
```

```python
import numpy as np

col = 'Emission_GES_kgCO2/m²/an'

data = df[col].dropna()


# Calcul des quantiles
q1 = np.quantile(df[col], 0.25)
q3 = np.quantile(df[col], 0.75)

# Calcul IQR
iqr = q3 - q1

# Bornes
borne_basse = q1 - 1.5 * iqr
borne_haute = q3 + 1.5 * iqr

# Affichage
print("=== IQR ===")
print(f"Q1 : {q1}")
print(f"Q3 : {q3}")
print(f"IQR : {iqr}")
print()
print(f"Borne basse : {borne_basse}")
print(f"Borne haute : {borne_haute}")

# Comptage des outliers
outliers = df[(df[col] < borne_basse) | (df[col] > borne_haute)]
print()
print(f"Nombre d'outliers : {len(outliers)}")
```
